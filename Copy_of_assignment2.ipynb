# PUSH THE CHANGES TO GITHUB

```python
import os
import subprocess
from subprocess import CalledProcessError # Import CalledProcessError
from google.colab import userdata

# Navigating to the repo directory
os.chdir(
'/content/colab-git-assignment2--K.O-'
)

token = userdata.get('GITHUB_TOKEN')
username ="kayodedada16-rgb"
repo ="colab-git-assignment2--K.O-"

remote_url =f"https://{username}:{token}@github.com/{username}/{repo}.git"

# Set remote URL with token authentication
print("Setting remote URL...")
set_url_result = subprocess.run(["git","remote","set-url","origin", remote_url], capture_output=True, text=True, check=False)
if set_url_result.returncode != 0:
    print("Error setting remote URL:")
    print("Stdout:", set_url_result.stdout)
    print("Stderr:", set_url_result.stderr)
    raise CalledProcessError(set_url_result.returncode, set_url_result.args, output=set_url_result.stdout, stderr=set_url_result.stderr)
print("Remote URL set successfully.")


# Pull remote changes with rebase to align histories
print("Attempting to pull changes with rebase...")
pull_result = subprocess.run(["git","pull","--rebase","origin","main"], capture_output=True, text=True, check=False)
if pull_result.returncode != 0:
    print("Error pulling with rebase:")
    print("Stdout:", pull_result.stdout)
    print("Stderr:", pull_result.stderr)
    raise CalledProcessError(pull_result.returncode, pull_result.args, output=pull_result.stdout, stderr=pull_result.stderr)
print("Pull with rebase successful.")


#  Push local commits to GitHub
print("Attempting to push local commits to GitHub...")
push_result = subprocess.run(["git","push","-u","origin","main"], capture_output=True, text=True, check=False)
if push_result.returncode != 0:
    print("Error pushing to GitHub:")
    print("Stdout:", push_result.stdout)
    print("Stderr:", push_result.stderr)
    raise CalledProcessError(push_result.returncode, push_result.args, output=push_result.stdout, stderr=push_result.stderr)

print("Successfully pushed to GitHub!")
```
